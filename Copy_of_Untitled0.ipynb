```python

from huggingface_hub import snapshot_download
import os

model_path = "/content/finbert"

snapshot_download(
    repo_id="ProsusAI/finbert",
    local_dir=model_path,
    allow_patterns=[
        "config.json",
        "model.safetensors",
        "tokenizer_config.json",
        "special_tokens_map.json",
        "vocab.txt"
    ]
)

print("FinBERT download completed!")
print(os.listdir(model_path))
```

```python

import shutil

shutil.make_archive(
    '/content/finbert',
    'zip',
    '/content/finbert'
)

print("FinBERT ZIP created successfully!")
```
